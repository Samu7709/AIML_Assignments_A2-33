#Conceptual Details

Intelligent Agent

An intelligent agent is a system that perceives its environment through sensors and performs actions through actuators to achieve a particular goal.

In this experiment:

Agent: Vacuum Cleaner Agent

Environment: School

The agent continuously observes the state of the rooms and performs actions based on the observed condition.

Environment States

Each room can have one of three states:

1.Clean

2.Dirty

3.Wet

4.Agent Actions

The agent can perform the following actions:

1.Move to Floor,

2.Move to Room,

3.Clean Dirty Room,

4.Dry/Clean Wet Room,

5.Do Nothing if Room is Clean.

#Algorithm

Algorithm: Vacuum Cleaner Agent

Step 1: Start the program.

Step 2: Create the school environment with two floors.

Step 3: Create four rooms on each floor.

Step 4: Randomly assign each room one of the following states:

Clean / Dirty / Wet

Step 5: Display the initial status of all rooms.

Step 6: Initialize the vacuum cleaner agent and set its score to zero.

Step 7: Move the agent to the first floor.

Step 8: Check each room on the floor.

Step 9: If the room is Dirty:

Clean the room.
Change its state to Clean.
Add +10 to the score.

Step 10: If the room is Wet:

Dry and clean the room.
Change its state to Clean.
Add +10 to the score.

Step 11: If the room is already Clean:

Take no cleaning action.
Continue to the next room.

Step 12: Move to the next floor and repeat the process.

Step 13: After checking all rooms, display the final performance score.

Step 14: Display the final status of the school.

Step 15: Stop.

```python
import random


# Environment
class School:
    def __init__(self):
        self.floors = {
            "Floor 1": {
                "Room 1": random.choice(["Dirty", "Clean", "Wet"]),
                "Room 2": random.choice(["Dirty", "Clean", "Wet"]),
                "Room 3": random.choice(["Dirty", "Clean", "Wet"]),
                "Room 4": random.choice(["Dirty", "Clean", "Wet"]),
            },
            "Floor 2": {
                "Room 1": random.choice(["Dirty", "Clean", "Wet"]),
                "Room 2": random.choice(["Dirty", "Clean", "Wet"]),
                "Room 3": random.choice(["Dirty", "Clean", "Wet"]),
                "Room 4": random.choice(["Dirty", "Clean", "Wet"]),
            }
        }

    def display_status(self):
        print("\nSchool Status:")

        for floor, rooms in self.floors.items():
            print(f"\n{floor}")

            for room, status in rooms.items():
                print(f"  {room}: {status}")


# Agent
class VacuumCleanerAgent:
    def __init__(self, school):
        self.school = school
        self.score = 0

    def clean_school(self):
        print("\nVacuum Cleaner Agent Started...\n")

        for floor, rooms in self.school.floors.items():

            print(f"\nMoving to {floor}")
            self.score -= 1  # Cost for changing floor

            for room, status in rooms.items():

                print(f"Checking {room}... ", end="")
                self.score -= 1  # Cost for moving to a room

                if status == "Dirty":
                    print("Dirty -> Cleaning")
                    self.school.floors[floor][room] = "Clean"
                    self.score += 10  # Reward for cleaning

                elif status == "Wet":
                    print("Wet -> Drying and Cleaning")
                    self.school.floors[floor][room] = "Clean"
                    self.score += 10  # Reward for cleaning

                else:
                    print("Already Clean")

        print("\nCleaning Completed!")
        print("Final Performance Score:", self.score)


# Main Program
school = School()

print("Initial Status:")
school.display_status()

agent = VacuumCleanerAgent(school)
agent.clean_school()

print("\nFinal Status:")
school.display_status()
```

## Result and Analysis

The Vacuum Cleaner Agent successfully operates in the school environment and identifies the condition of each room. The agent takes appropriate actions based on the condition of the room.

If a room is **Dirty**, the agent cleans it and changes its status to **Clean**. If a room is **Wet**, the agent dries and cleans it, changing its status to **Clean**. If a room is already **Clean**, the agent takes no action.

After completing the cleaning process, **all rooms in the school become Clean**. The agent also maintains a performance score based on the rewards received for cleaning and the costs incurred while moving between floors and rooms.

For example, if the agent cleans **5 rooms**, the cleaning reward is:

**Cleaning Reward = 5 × 10 = 50**

If the agent incurs **10 movement/checking costs**, the movement cost is:

**Movement Cost = 10 × (−1) = −10**

Therefore, the final performance score is:

**Final Score = 50 − 10 = 40**

The exact final score may vary in each execution because the initial condition of every room is generated randomly. The experiment demonstrates that the agent can effectively perceive the environment, make decisions, perform cleaning actions, and achieve its goal of maintaining a clean school environment.
